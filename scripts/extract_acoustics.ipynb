```python
import tdt
import numpy as np
import soundfile as sf
import matplotlib.pyplot as plt
```

```python
# import tdt tank
data = tdt.read_block('..\\data\\allie\\tank')
```

```python
mic = data.streams.Mic1
```

```python
mic_data = mic.data
mic_fs = mic.fs
# scale data to be between -1 and 1
mic_data_scaled = mic_data / np.max(np.abs(mic_data))
```

```python
# plot the mic
plt.figure(figsize=(10, 4))
plt.plot(mic_data_scaled)
plt.title('Microphone Signal')
plt.xlabel('Sample Index')
```

```python
# package the data into a dictionary
mic_dict = {
    'data': mic_data_scaled,
    'fs': mic_fs
}
# export to wav
sf.write('../data/allie/mic1.wav', mic_data_scaled, int(mic_fs))
```
